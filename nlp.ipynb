```python
%pip install nltk
import pandas as pd
import nltk
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

# Download tokenizer (run once)
nltk.download("punkt")

# Load dataset
df = pd.read_csv("Tweets.csv")

# Initialize the Porter Stemmer
stemmer = PorterStemmer()

# Function to apply stemming
def stem_text(text):
    words = word_tokenize(text)  # Tokenize text into words
    stemmed_words = [stemmer.stem(word) for word in words]  # Apply stemming
    return " ".join(stemmed_words)  # Reconstruct the sentence

# Apply stemming to the text column
df["stemmed_text"] = df["text"].astype(str).apply(stem_text)

# Display the first few rows
print(df[["text", "stemmed_text"]].head())
```

```python
import pandas as pd
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

# Load the dataset
df = pd.read_csv("Tweets.csv")

# Initialize the Porter Stemmer
stemmer = PorterStemmer()

# Function to apply stemming to text
def stem_text(text):
    if isinstance(text, str):  # Ensure it's a string
        words = word_tokenize(text)  # Tokenize words
        return [stemmer.stem(word) for word in words]  # Apply stemming and return a list of words
    return []

# Apply stemming to the "text" column
df["stemmed_words"] = df["text"].apply(stem_text)

# Create a separate variable for only stemmed data
stemmed_data = df["stemmed_words"]

# Display sample output
print(stemmed_data.head())

# Save the processed dataset
df.to_csv("Tweets_stemmed.csv", index=False)
```

```python
stm=[" ".join(i) for i in stemmed_data]
```
