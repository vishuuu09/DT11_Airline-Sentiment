```python

# Initialize the Porter Stemmer
stemmer = PorterStemmer()

# Function to perform stemming
def apply_stemming(text):
    # Convert to lowercase and remove punctuation
    text = text.lower().translate(str.maketrans('', '', string.punctuation))
    
    # Tokenize words
    words = text.split()  # Using simple split instead of word_tokenize to avoid missing dependencies
    
    # Apply stemming
    stemmed_words = [stemmer.stem(word) for word in words]
    
    return " ".join(stemmed_words)

# Apply stemming to the first 5 tweets
df["stemmed_text"] = df["text"].dropna().head(5).apply(apply_stemming)

# Display results
df[["text", "stemmed_text"]].head(5)
```
